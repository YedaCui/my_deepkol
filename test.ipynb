```python
import torch

# 创建一个输入张量，并设置 requires_grad=True 以便追踪其梯度
x = torch.tensor([2.0], requires_grad=True)

# 定义一个函数 y = x^2
y = x ** 2

# 执行反向传播以计算梯度
y.backward()

# x.grad 现在包含了 dy/dx 的值
print(x.grad)  # 输出: tensor([4.])
```
